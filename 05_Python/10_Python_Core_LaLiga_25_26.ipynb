```python
# Creamos una lista vacía para almacenar equipos locales.
home_teams = []

# Recorremos todos los partidos.
for match in matches:

    # Obtenemos los equipos locales.
    home_team = match["HomeTeam"]

    # Lo almacenamos en la lista.
    home_teams.append(home_team)

# Creamos una lista vacía para almacenar equipos visitantes.
away_teams = []

# Recorremos todos los partidos.
for match in matches:

    # Obtenemos los equipos visitantes.
    away_team = match["AwayTeam"]

    # Almacenamos los equipos en la lista.
    away_teams.append(away_team)

# Unimos las dos listas
all_teams = home_teams + away_teams

# Eliminamos los duplicados
unique_teams = set(all_teams)
print("Número de equipos:", len(unique_teams))
print("\nEquipos:",unique_teams)
```

```python
# Creamos un diccionario vacío para almacenar el númeor de partidos de cada equipo.
matches_by_team = {}

# Recorremos los partidos.
for match in matches:

    # Obtenemos el nombre del equipo local.
    home_team = match["HomeTeam"]

    # Obtenemos el nombre del equipo vsitante.
    away_team = match["AwayTeam"]

    # Comprobamos si el equipo está en el diccionario.
    if home_team not in matches_by_team:

        # Si no está, lo añadimos.
        matches_by_team[home_team] = 1

    else:

        # Si existe, sumamos el partido.
        matches_by_team[home_team] += 1

    # Comprobamos si el equipo visitante está en el diccionario.
    if away_team not in matches_by_team:

        # Si no existe, lo añadimos.
        matches_by_team[away_team] = 1

    else:

        # Si existe, sumamos el partido.
        matches_by_team[away_team] += 1

# Mostramos el número de partidos por equipo.
for team in matches_by_team:
    print(team,":",matches_by_team[team])
```

```python
# Creamos un diccionario vacío para almacenar los goles por cada equipo.
goals_for = {}

# Recorremos los partidos.
for match in matches:

    # Obtenemos los equipos locales.
    home_team = match["HomeTeam"]

    # Obtenemos los equipos visitantes.
    away_team = match["AwayTeam"]

    # Obtenemos goles locales y convertimos en entero.
    home_goals = int(match["FTHG"])

    # Obtenemos goles visitantes y convertimos en entero.
    away_goals = int(match["FTAG"])

    # Comprobamos que el equipo local esté en el diccionario.
    if home_team not in goals_for:

        # Si no está, lo añadimos.
        goals_for[home_team] = home_goals

    else:

        # Si está, lo sumamos.
        goals_for[home_team] += home_goals

    # Comprobamos que el equipo visitante no esté en el diccionario.
    if away_team not in goals_for:

        # Si no está, lo añadimos.
        goals_for[away_team] = away_goals

    else:

        # Si está, lo sumamos.
        goals_for[away_team] += away_goals

for team in goals_for:
    print(team,":",goals_for[team])
    
    
    
```

```python
# Creamos otro diccionario vacío para los goles en contra.
goals_against = {}

# Recorremos todos los partidos.
for match in matches:

    # Obtenemos los equipos locales.
    home_team = match["HomeTeam"]

    # Obtenemos los equipops visitantes.
    away_team = match["AwayTeam"]

    # Obtenemos goles en contra de equipo local (cambiamos el concepto de los goles respecto al bloque anterior).
    home_goals_against = int(match["FTAG"])

    # Obtenemos los goles en contra del equipo visitante.
    away_goals_against = int(match["FTHG"])

    # Comprobamos que el equipo local esté en el diccionario.
    if home_team not in goals_against:

        # Si no está, lo añadimos.
        goals_against[home_team] = home_goals_against

    else:

        # Si existe, lo sumamos.
        goals_against[home_team] += home_goals_against

    # Procedemos igual con los equipos visitantes.
    if away_team not in goals_against:

        goals_against[away_team] = away_goals_against

    else:

        goals_against[away_team] += away_goals_against

for team in goals_against:
    print(team,":",goals_against[team])
```

```python
# Creamos un diccionario para victorias.
wins = {}

# Creamos un diccionrio para empates.
draws = {}

# Creamos un diccionario para derrotas.
losses = {}

# Recorremos los partidos.
for match in matches:

    # Obtenemos equipo local.
    home_team = match['HomeTeam']

    # Obtenemos equipo visitante.
    away_team = match['AwayTeam']

    # Obtenemos resultado final.
    result = match['FTR']

    # Si gana el equipo local.
    if result == 'H':

        # Registramos la victoria del equipo local.
        if home_team not in wins:
            wins[home_team] = 1
        else:
            wins[home_team] += 1

        # Registramos la derrota del equipo visitante.
        if away_team not in losses:
            losses[away_team] = 1
        else:
            losses[away_team] += 1

    # Si hay empate.
    elif result == 'D':

        # Registramos el empate del equipo local.
        if home_team not in draws:
            draws[home_team] = 1
        else:
            draws[home_team] += 1

        # Registramos el empate del equipo visitante.
        if away_team not in draws:
            draws[away_team] = 1
        else:
            draws[away_team] += 1

    # Si gana equipo visitante.
    elif result == 'A':

        # Registramos la derrota del quipo local.
        if home_team not in losses:
            losses[home_team] = 1
        else:
            losses[home_team] += 1

        # Registramos la vistoria del quipo visitante.
        if away_team not in wins:
            wins[away_team] = 1
        else:
            wins[away_team] += 1

# Mostramos las victorias de cada equipo.
for team in wins:
    print(team, ':', wins[team])
    
```

```python
# Creamos un diccionario para almacenar los puntos por equipo.
points = {}

# Recorremos los equipos.
for team in teams:

    # Obtenemos las victorias del equipo.
    team_wins = wins.get(team,0)

    # Obtenemos los empates del equipo.
    team_draws = draws.get(team,0)

    # Calculamos los puntos.
    team_points = (team_wins*3) + team_draws

    # Guardamos los datos en el diccionario.
    points[team] = team_points

# Mostramos los puntos de cada equipo.
for team in points:
    print(team,":",points[team])
```

```python
# Para mostrarlo ordenado.
for team in sorted(points, key=points.get, reverse = True):
    print(team,":",points[team])
```

```python
# Creamos un diccionario para almacenar la diferencia de goles de cada equipo.
goal_difference = {}

# Recorremos los equipos.
for team in teams:

    # Obtenemos goles a favor.
    team_goals_for = goals_for.get(team,0)

    # Obtenemos goles en contra.
    team_goals_against = goals_against.get(team,0)

    # Calculamos la diferencia de goles.
    team_goal_difference = team_goals_for - team_goals_against

    # Guardamos en el diccionario los resultados.
    goal_difference[team] = team_goal_difference

# Mostramos los resultados.
for team in goal_difference:
    print(team,":",goal_difference[team])
```

```python
# Ordenamos resultados.
for team in sorted(goal_difference,key=goal_difference.get,reverse=True):
    print(team,":",goal_difference[team])
```

```python
# Creamos un diccionario vacío para almacenar toda la clasificación.
classification = {}

# Recorremos los equipos.
for team in teams:

    # Creamos un diccionario con las estadísticas del equipo.
    team_stats = {
        'matches':matches_by_team.get(team,0),
        'wins':wins.get(team,0),
        'draws':draws.get(team,0),
        'losses':losses.get(team,0),
        'goals_for':goals_for.get(team,0),
        'goals_against':goals_against.get(team,0),
        'goal_difference':goal_difference.get(team,0),
        'points':points.get(team,0),
    }

    # Guardamos las estadísticas en el primer diccionario.
    classification[team] = team_stats

# Comprobamos el número de equipos en el diccionario.
print(len(classification))
```

```python
# Ordenamos los equipos seún sus puntos.
ranking = sorted(
    classification,
    key=lambda team: classification[team]["points"],
    reverse=True
)

# Recorremos los equipos en el orden de la clasificación
for position, team in enumerate (ranking, start=1):

    # Mostramos posición, equipo y puntos.
    print(
        position,
        ".",
        team,
        ":",
        classification[team]["points"],
        "puntos"
    )
```

```python
# lambda
# Una lambda es una función pequeña y anónima.
# Se utiliza cuando necesitamos una función sencilla
# y no queremos crearla mediante def.
# Ejemplo:
# lambda team: classification[team]["points"]
# Se puede interpretar como:
# "para cada team, devuelve sus puntos".
```

```python
# enumerate()
# enumerate() permite recorrer una colección obteniendo
# simultáneamente la posición y el elemento.
# Ejemplo:
# for position, team in enumerate(ranking, start=1):
# position → posición
# team → elemento actual
# start=1 hace que la numeración empiece en 1.
```

```python
# Ordenamos la clasificación utilizando dos criterios, primero puntos y después diferencia de goles.
ranking = sorted(
    classification,
    key=lambda team:(
    classification[team]['points'],
    classification[team]['goal_difference']
        ),
    reverse=True
)

# Mostramos la clasificación ordenada.
for position, team in enumerate(ranking, start=1):
    print(
        position,
        ".",
        team,
        ":",
        classification[team]['points'],
        "puntos | DG:",
        classification[team]['goal_difference']
            )
```

```python
# sorted() puede utilizar varios criterios de ordenación.
#
# Para ello podemos hacer que la función key devuelva
# una tupla con varios valores.
#
# Ejemplo:
#
# key=lambda team: (
#     classification[team]["points"],
#     classification[team]["goal_difference"]
# )
#
# Python compara primero el primer elemento de la tupla.
# Si hay empate, compara el segundo.
#
# Con reverse=True:
# - puntos → de mayor a menor
# - diferencia de goles → de mayor a menor
```

```python
# Mostramos el encabezado de la clasificación.
print("POS | EQUIPO | PJ | G | E | P | GF | GC | DG | PTS")

# Recorremos los equipos en el orden de clasificación.
for position, team in enumerate(ranking, start=1):
    
    # Obtenemos partidos jugados.
    matches_played = classification[team]['matches']

    # Obtenemos las victorias.
    team_wins = classification[team]['wins']

    # Obtenemos los empates.
    team_draws = classification[team]["draws"]

    # Obtenemos las derrotas.
    team_losses = classification[team]["losses"]

    # Obtenemos los goles a favor.
    team_goals_for = classification[team]["goals_for"]

    # Obtenemos los goles en contra.
    team_goals_against = classification[team]["goals_against"]

    # Obtenemos la diferencia de goles.
    team_goal_difference = classification[team]["goal_difference"]

    # Obtenemos los puntos.
    team_points = classification[team]["points"]

    # Mostramos toda la información del equipo.
    print(
    position,
    ".",
    team,
    "| PJ:", matches_played,
    "| G:", team_wins,
    "| E:", team_draws,
    "| P:", team_losses,
    "| GF:", team_goals_for,
    "| GC:", team_goals_against,
    "| DG:", team_goal_difference,
    "| PTS:", team_points
    ) 
```

```python
# ACUMULADORES
#
# Un acumulador permite ir almacenando un resultado
# mientras recorremos una colección.
#
# Primero inicializamos la variable:
# total = 0
#
# Después recorremos los datos:
# for elemento in datos:
#
# Y vamos acumulando:
# total += valor
#
# Esto equivale a:
# total = total + valor
```

```python
# PROMEDIOS
#
# Para calcular un promedio podemos dividir el total
# acumulado entre el número de elementos.
#
# Ejemplo:
#
# promedio = total / len(datos)
#
# round(valor, 2) permite redondear el resultado
# a dos decimales.
```

```python
# Creamos un contador para victorias locales.
home_wins = 0

# Creamos un contador para empates.
draws_by_team = 0

# Creamos un contador para victorias visitantes.
away_wins = 0

# Recorremos los partidos.
for match in matches:

    # Comprobamos el resultado de los partidos.
    if match ['FTR'] == 'H':

        # Sumamos una victoria local.
        home_wins += 1

    elif match['FTR'] == 'D':

        # Sumamos un empate.
        draws_by_team += 1

    elif match['FTR'] == 'A':

        # Sumamos victoria visitante
        away_wins += 1

# Mostramos los resultados.
print("Victorias locales:", home_wins)
print("Empates:", draws_by_team)
print("Victorias visitantes:", away_wins)
```

```python
# Comprobamos que la suma de los tres resultados
# coincide con el número total de partidos.
print(home_wins + draws_by_team + away_wins)
```

```python
# CONTADORES
#
# Un contador permite contar cuántas veces se cumple
# una determinada condición.
#
# Primero inicializamos:
# contador = 0
#
# Después recorremos los datos:
# for elemento in datos:
#
# Comprobamos una condición:
# if condicion:
#
# Y aumentamos el contador:
# contador += 1
#
# Diferencia con un acumulador:
#
# CONTADOR:
# suma normalmente 1 cada vez que se cumple una condición.
#
# ACUMULADOR:
# suma valores que pueden ser diferentes en cada iteración.
```

```python
# Calculamos porcentaje de victorias locales.
home_wins_percentage = round(home_wins/len(matches)*100,2)

# Calculamos porcentaje de empates.
draws_percentage = round(draws_by_team/len(matches)*100,2)

# Calculamos porcentaje de victorias visitante.
away_wins_percentage = round(away_wins/len(matches)*100,2)

# Mostramos resultados.
print("Porcentaje de victorias locales:",home_wins_percentage,"%")
print("Porcentaje de empates:",draws_percentage,"%")
print("Porcentaje de victorias visitantes:",away_wins_percentage,"%")
```

```python
# ACUMULAR DATOS POR CATEGORÍAS
#
# Podemos utilizar diferentes acumuladores para separar
# los datos en diferentes grupos.
#
# Ejemplo:
#
# home_goals_total = 0
# away_goals_total = 0
#
# Después, dentro del bucle:
#
# home_goals_total += home_goals
# away_goals_total += away_goals
#
# De esta forma mantenemos separados los valores
# correspondientes a cada categoría.
```

```python
# Creamos un dccionario vacío para almacenar cuantos partidos tuvieron cada número de goles.
goals_distribution = {}

# Recorreomos los partidos
for match in matches:

    # Obtenemos goles del equipo local.
    home_goals = int(match['FTHG'])

    # Obtenemos goles visitantes.
    away_goals = int(match['FTAG'])

    # Obtenemos el total de goles.
    total_match_goals = home_goals + away_goals

    # Comprobamos si existe el número en el diccionario.
    if total_match_goals in goals_distribution:

        # Si ya existe, aumentamos el contador en 1.
        goals_distribution[total_match_goals] += 1

    else:

        # Si es la primera vez que aparece, creamos el registro.
        goals_distribution[total_match_goals] = 1

# Mostramos la distribución de los goles.
print(goals_distribution)
```

```python
# Ordenamos los resultados.
for goals in sorted(goals_distribution):
    print(goals,"goles:",goals_distribution[goals],"partidos")
```

```python
# Sumamos el número de partidos de todas las categorías.
total_distribution = sum(goals_distribution.values())

# Mostramos el resultado.
print("Partidos contabilizados:", total_distribution)
```

```python
def filtrar_partidos(team):

    # Creamos una lista vacía para almacenar partidos.
    filtered_matches = []

    # Recorremos los partidos.
    for match in matches:

        # Obtenemos el equipo local.
        home_team = match['HomeTeam']

        # Obtenemos el equipo visitante.
        away_team = match['AwayTeam']

        # Comprobamos si el equipo participa en el partido.
        if home_team == team or away_team == team:

            # Añadimos a la lista.
            filtered_matches.append(match)
    
    # Devolvemos los partidos encontrados.
    return filtered_matches
```

```python
def analizar_equipo(team):

    # Creamos un diccionario con las estadísticas del equipo.
    stats = {
        "matches": len(filtrar_partidos(team)),
        "wins": wins[team],
        "draws": draws[team],
        "losses": losses[team],
        "goals_for": goals_for[team],
        "goals_against": goals_against[team],
        "goal_difference": goals_for[team] - goals_against[team],
        "points": points[team]
    }

    # Devolvemos las estadísticas.
    return stats
```
